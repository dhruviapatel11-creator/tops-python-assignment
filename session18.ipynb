```python
import re

text = "Contacts: +91-9876543210, +91-8765432109, +91-123456789"

pattern = r"\+91-\d{10}"

phone_numbers = re.findall(pattern, text)

print(phone_numbers)
```

```python
import re

def check_date(text):
    pattern = r"\b\d{2}/\d{2}/\d{4}\b"
    return bool(re.search(pattern, text))

print(check_date("My birthday is 23/10/2005"))
print(check_date("My birthday is 23-10-2005"))
```

```python
import re

text = "Prices: Rs. 299, Rs. 1500, Rs. 750, Rs. 200"

prices = re.findall(r"Rs\.?\s*(\d+)", text)

prices = [int(price) for price in prices]

print("Prices:", prices)
print("Total:", sum(prices))
```

```python
import re

text = "Contact abc@gmail.com or xyz@yahoo.com for details."

pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"

result = re.sub(pattern, "[hidden email]", text)

print(result)
```

```python
import re

with open("comments.txt", "r", encoding="utf-8") as file:
    text = file.read()

pattern = r"(?<![A-Za-z0-9_])@[A-Za-z0-9_]{2,29}[A-Za-z0-9](?![A-Za-z0-9_])"

usernames = re.findall(pattern, text)

unique_usernames = sorted(set(usernames))

print("Unique Instagram usernames:")
for username in unique_usernames:
    print(username)
```
