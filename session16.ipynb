```python
def playlist_generator(songs):
    for song in songs:
        yield song

songs = ["Perfect", "Shape of You", "Blinding Lights", "Levitating"]

for song in playlist_generator(songs):
    print(song)
```

```python
cart = ["Pizza", "Burger", "Fries", "Coke"]

for index, item in enumerate(cart, start=1):
    print(index, item)
```

```python
teams = ["Mumbai Indians", "Chennai Super Kings", "Royal Challengers Bengaluru"]
points = [18, 16, 14]

for team, point in zip(teams, points):
    print(f"Team: {team}, Points: {point}")
```

```python
def order_id_generator():
    order_id = 1001

    while True:
        yield order_id
        order_id += 1

y = order_id_generator()

print(next(y))
print(next(y))
print(next(y))
print(next(y))
print(next(y))
```
