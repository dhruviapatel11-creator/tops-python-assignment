```python
class Playlist:
    def __init__(self):
        self.__songs = []

    def add_song(self, song):
        self.__songs.append(song)

    def show_playlist(self):
        print("Playlist:", self.__songs)


my_playlist = Playlist()

my_playlist.add_song("Perfect")
my_playlist.add_song("Shape of You")
my_playlist.add_song("Blinding Lights")

my_playlist.show_playlist()
```

```python
class Product:
    def __init__(self, price):
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, price):
        self.__price = price


product1 = Product(1500)

print("Original Price: ₹", product1.get_price())

product1.set_price(1200)

print("Updated Price: ₹", product1.get_price())
```

```python
class Movie:
    def __init__(self, rating):
        self.__rating = 0
        self.set_rating(rating)

    def get_rating(self):
        return self.__rating

    def set_rating(self, rating):
        if 0 <= rating <= 10:
            self.__rating = rating
        else:
            print("Error: Rating must be between 0 and 10.")


movie1 = Movie(8.5)
print("Movie Rating:", movie1.get_rating())

movie1.set_rating(11)
print("Movie Rating:", movie1.get_rating())
```

```python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class Paytm(PaymentMethod):
    def pay(self, amount):
        print("Paying ₹", amount, "using Paytm")


class PhonePe(PaymentMethod):
    def pay(self, amount):
        print("Paying ₹", amount, "using PhonePe")


payment1 = Paytm()
payment2 = PhonePe()

payment1.pay(500)
payment2.pay(800)
```
