```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price * 0.90


product1 = Product("Shirt", 1000)

print("Product:", product1.name)
print("Discounted Price: ₹", product1.get_discounted_price())
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price * 0.90


class Electronics(Product):
    def get_discounted_price(self):
        return self.price * 0.80


product1 = Product("Shirt", 1000)
product2 = Electronics("Headphones", 2000)

print("Product Discounted Price: ₹", product1.get_discounted_price())
print("Electronics Discounted Price: ₹", product2.get_discounted_price())
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price * 0.90


class Electronics(Product):
    def get_discounted_price(self):
        return self.price * 0.80


def show_final_price(item):
    print("Product:", item.name)
    print("Discounted Price: ₹", item.get_discounted_price())


product1 = Product("Shirt", 1000)
product2 = Electronics("Laptop", 50000)

show_final_price(product1)
show_final_price(product2)
```

```python
class Ticket:
    def __init__(self, price):
        self.price = price

    def get_final_price(self):
        return self.price


class PremiumTicket(Ticket):
    def get_final_price(self):
        return super().get_final_price() + 50


ticket1 = Ticket(200)
ticket2 = PremiumTicket(300)

print("Regular Ticket Price: ₹", ticket1.get_final_price())
print("Premium Ticket Price: ₹", ticket2.get_final_price())
```
