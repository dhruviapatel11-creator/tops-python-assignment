```python
#1
songs = ['Shape Of You', 'Blinding Lights', 'Levitating', 'Senorita']

cleaned_songs = list(map(lambda song: song.lower(), songs))

print(cleaned_songs)
```

```python
#2
ratings = [4.2, 3.8, 4.5, 2.9, 3.5]

high_ratings = list(filter(lambda rating: rating > 4.0, ratings))

print(high_ratings)

```

```python
#3
from functools import reduce

prices = [499, 1299, 299, 799]

total = reduce(lambda a, b: a + b, prices)
print("Total Price:", total)
```

```python
#4
def format_followers(count):
    if count >= 1000000:
        return f"{count / 1000000:g}M"
    elif count >= 1000:
        return f"{count / 1000:g}K"
    else:
        return str(count)


followers = [950, 1500, 25000, 1200000]

formatted = list(map(format_followers, followers))

print(formatted)
```

```python
#5
scores = [101, 98, 120, 77, 88]

even_scores = list(filter(lambda score: score % 2 == 0, scores))

print(even_scores)
```
