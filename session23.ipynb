```python
import requests

api_key = "YOUR_API_KEY"
city = "Ahmedabad"

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": city,
    "appid": api_key,
    "units": "metric"
}

try:
    response = requests.get(url, params=params, timeout=10)
    response.raise_for_status()

    data = response.json()

    print("City:", data["name"])
    print("Temperature:", data["main"]["temp"], "°C")
    print("Weather:", data["weather"][0]["description"])

except requests.RequestException as e:
    print("Unable to fetch weather:", e)
```

```python
import requests
from datetime import datetime

url = "https://api.coingecko.com/api/v3/simple/price"

params = {
    "ids": "bitcoin,ethereum",
    "vs_currencies": "usd"
}

try:
    response = requests.get(url, params=params, timeout=10)
    response.raise_for_status()

    data = response.json()

    print("Date and Time:", datetime.now().strftime("%d-%m-%Y %H:%M:%S"))
    print("Bitcoin Price: $", data["bitcoin"]["usd"])
    print("Ethereum Price: $", data["ethereum"]["usd"])

except requests.RequestException as e:
    print("Unable to fetch cryptocurrency prices:", e)
```

```python
import requests

api_key = "DEMO_KEY"

url = "https://api.nasa.gov/planetary/apod"
params = {"api_key": api_key}

try:
    response = requests.get(url, params=params, timeout=15)
    response.raise_for_status()

    data = response.json()

    print("Title:", data["title"])
    print("Explanation:", data["explanation"])

    if data["media_type"] == "image":
        image_url = data["url"]

        image_response = requests.get(image_url, timeout=20)
        image_response.raise_for_status()

        with open("nasa_apod.jpg", "wb") as file:
            file.write(image_response.content)

        print("Image saved as nasa_apod.jpg")
    else:
        print("Today's APOD is not an image.")

except requests.RequestException as e:
    print("Unable to fetch NASA data:", e)
```

```python
import requests

url = "https://disease.sh/v3/covid-19/countries/India"

try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()

    data = response.json()

    print("-" * 55)
    print("COVID-19 Statistics - India")
    print("-" * 55)
    print(f"{'Metric':<20}{'Count':>20}")
    print("-" * 40)
    print(f"{'Total Cases':<20}{data['cases']:>20,}")
    print(f"{'Recovered':<20}{data['recovered']:>20,}")
    print(f"{'Deaths':<20}{data['deaths']:>20,}")
    print("-" * 40)

except requests.RequestException as e:
    print("Unable to fetch COVID-19 data:", e)
```

```python
import requests

weather_url = "https://api.openweathermap.org/data/2.5/weather"
crypto_url = "https://api.coingecko.com/api/v3/simple/price"

api_key = "YOUR_API_KEY"

# Fetch Mumbai weather
try:
    weather_response = requests.get(
        weather_url,
        params={
            "q": "Mumbai",
            "appid": api_key,
            "units": "metric"
        },
        timeout=10
    )
    weather_response.raise_for_status()

    weather_data = weather_response.json()
    temperature = weather_data["main"]["temp"]
    weather_result = f"Mumbai Temperature: {temperature} °C"

except (requests.RequestException, KeyError, ValueError):
    weather_result = "Weather data is currently unavailable."

# Fetch Bitcoin price
try:
    crypto_response = requests.get(
        crypto_url,
        params={
            "ids": "bitcoin",
            "vs_currencies": "usd"
        },
        timeout=10
    )
    crypto_response.raise_for_status()

    crypto_data = crypto_response.json()
    bitcoin_price = crypto_data["bitcoin"]["usd"]
    crypto_result = f"Bitcoin Price: ${bitcoin_price:,.2f}"

except (requests.RequestException, KeyError, ValueError):
    crypto_result = "Bitcoin price is currently unavailable."

# Display both results
print("===== Current Information =====")
print(weather_result)
print(crypto_result)
```
