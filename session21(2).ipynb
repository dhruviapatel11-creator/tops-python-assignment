```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title": "My First Post",
    "body": "Learning Python APIs",
    "userId": 1
}

response = requests.post(url, json=data)

print("Status Code:", response.status_code)
print("Response Data:", response.json())
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

playlist_name = input("Enter playlist name: ")
description = input("Enter playlist description: ")

data = {
    "title": playlist_name,
    "body": description,
    "userId": 1
}

response = requests.post(url, json=data)
result = response.json()

print("Playlist ID:", result["id"])
```

```python
import requests

url = "https://reqres.in/api/users"

data = {
    "username": "dhruvi",
    "job": "Python Student"
}

headers = {
    "x-api-key": "reqres-free-v1"
}

response = requests.post(url, json=data, headers=headers)

print("Status Code:", response.status_code)

result = response.json()

if response.ok:
    print("User ID:", result["id"])
    print("Created At:", result["createdAt"])
else:
    print("Request failed:", result)
```

```python
import requests
import csv

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)
posts = response.json()

latest_posts = posts[-5:]

with open("posts.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Title", "UserID"])

    for post in latest_posts:
        writer.writerow([post["title"], post["userId"]])

print("Data saved to posts.csv")
```

```python
import requests
import json

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)
posts = response.json()

latest_posts = posts[-5:]

with open("posts.json", "w", encoding="utf-8") as file:
    json.dump(latest_posts, file, indent=4)

print("Data saved to posts.json")
```
