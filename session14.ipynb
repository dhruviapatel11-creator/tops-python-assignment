```python
# Create a text file and write 5 song names
with open("playlist.txt", "w") as file:
    file.write("Perfect\n")
    file.write("Kesariya\n")
    file.write("Heeriye\n")
    file.write("Tum Hi Ho\n")
    file.write("Apna Bana Le\n")

print("Songs written successfully!")
```

```python
# Read the playlist file
with open("playlist.txt", "r") as file:
    songs = file.readlines()

# Display each song in uppercase
for song in songs:
    print(song.strip().upper())
```

```python
import csv

# Read the CSV file
with open("ipl_scores.csv", "r", newline="") as file:
    reader = csv.DictReader(file)

    for match in reader:
        print("Match", match["match_id"], "- Winner:", match["winner"])
```

```python
import json

# Load the JSON file
with open("movies.json", "r") as file:
    movies = json.load(file)

# Display movie titles and ratings
for movie in movies:
    print("Title:", movie["title"], "| Rating:", movie["rating"])
```

```python
import json
from pathlib import Path

# Define the file path
file_path = Path("my_fav_apps.json")

# Check if the file exists
if not file_path.exists():
    apps = [
        {"name": "Instagram", "category": "Social Media"},
        {"name": "Zomato", "category": "Food Delivery"},
        {"name": "Paytm", "category": "Finance"}
    ]

    # Write data to the JSON file
    with open(file_path, "w") as file:
        json.dump(apps, file, indent=4)

    print("my_fav_apps.json created successfully!")
else:
    print("my_fav_apps.json already exists.")
```
