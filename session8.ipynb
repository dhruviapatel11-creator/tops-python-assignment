```python
steps = [4500, 7800, 9500, 12000, 10500, 8000, 11000]
days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

i = 0

while i < len(steps):
    if steps[i] > 10000:
        print("First day:", days[i])
        print("Steps:", steps[i])
        break
    i += 1
```

```python
def print_teams(teams):
    for team in teams:
        if len(team) <= 6:
            continue
        print(team)

ipl_teams = ["CSK", "Mumbai Indians", "RCB",
             "Rajasthan Royals", "Punjab Kings"]

print_teams(ipl_teams)
```

```python
durations = [210, 185, 240, 195, 220]

for position, duration in enumerate(durations, start=1):
    print(f"Song {position}: {duration} seconds")
```

```python
prices = [500, 0, 750, 400, 600, 300]
total = 0

for price in prices:
    if price == 0:
        continue
    if total + price > 2000:
        break  
    total +=price

print("Final Total: ₹", total)   
```
